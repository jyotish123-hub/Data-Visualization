```
import pandas as pd
import seaborn as sns
df=sns.load_dataset("titanic")
df.tail(3)
```

```
df.groupby("sex").size()
```

```
df.groupby("sex").count()
```

```
df.groupby("sex")["age"].mean()
```

```
df.groupby("sex")["age"].max()
```

```
df.groupby("sex")["age"].agg(["min","max","mean","std","var"])
```

```
df.groupby("sex")[["age","fare"]].agg(["min","max","mean","std","var"])
```

```
df.groupby("sex")[["age","fare"]].min()
```

```
df.groupby("sex")[["age","fare"]].agg({"age":"min","fare":"max"})
```

```
df.groupby("pclass").size()
```

```
df.groupby(["pclass","sex"])["fare"].max()
```

```
df.groupby(["pclass","sex"]).agg({"fare":"max"})
```

```
df.groupby(["pclass","sex"])["survived"].size()
```
